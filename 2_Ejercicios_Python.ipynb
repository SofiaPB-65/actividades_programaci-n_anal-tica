```python
#1. Crear una lista de notas
# Complete la lista con cinco notas y muestre la cantidad de elementos.
notas = [5.0, 4.5, 4.0, 3.0, 2.5]
print(len(notas))
print(notas)
```

```python
#2. Acceder por índice
# Use índices para imprimir la primera, la tercera y la última temperatura.
temperaturas = [18, 20, 19, 21, 22]
print(temperaturas[0])
print(temperaturas[2])
print(temperaturas[-1])
```

```python
#3. Actualizar un elemento
# Corrija la segunda venta, que fue registrada como 80 pero debía ser 85.
ventas = [120, 80, 200, 50]
ventas[1] = 85
print(ventas)
```

```python
#4. Agregar y extraer datos
# Agregue una nueva nota y extraiga la última nota registrada.
notas = [3.5, 4.0, 2.8]
notas.append(4.6)
ultima = notas.pop()
print(ultima)
print(notas)
```

```python
#5. Calcular promedio
# Complete el acumulador para calcular el promedio de una lista.
notas = [4.2, 3.8, 5.0, 2.9]
total = 0
for nota in notas:
    total = total + nota
    promedio = total / len(notas)
print(round(promedio, 2))
```

```python
#6. Conbtar aprobados
# Cuente cuántas notas son mayores o iguales a 3.0.
notas = [4.2, 2.5, 3.0, 1.8, 4.7]
aprobados = 0
for nota in notas:
    if nota >= 3.0:
        aprobados = aprobados + 1
print(aprobados)
```

```python
#7. Filtrar valore válidos
# Construya una lista con las temperaturas que estén entre -10 y 50 grados.
lecturas = [18, 200, 21, -99, 19, 22]
validas = [range(-10,50)]
for t in lecturas:
    if t >= -10 and t <= 50:
        validas.append(t)
print(validas)
```

```python
#8. Ordenar sin perder el original
# Use sorted() para crear una lista ordenada sin modificar la lista inicial.
montos = [120000, 85000, 210000, 50000]
ordenados = sorted(montos)
print(montos)
print(ordenados)
```

```python
#9.Comprensión de listas
# Complete la comprensión para obtener los cuadrados de los números pares.
numeros = [1, 2, 3, 4, 5, 6]
cuadrados_pares = [n**2 for n in numeros if n % 2 == 0]  #Operador de módulo (%) -Evalúa si el residuo de dividir n entre 2 es exactamente 0, identifica los números pares.
print(cuadrados_pares)
```

```python
#10.Lista animada
# Complete el doble ciclo para sumar todos los elementos de una matriz.
matriz = [[1, 2, 3], [4, 5, 6]]
total = 0
for fila in matriz:
    for valor in fila:    #El 1er ciclo recorre cada sublista (lista)
        total += valor
print(total)

## Por qué una matriz puede modelarse como una lista de listas??
# Una matriz se modela como una lista de listas porque cada fila
# es una lista individual contenida dentro de una lista principal.
```

```python
#15. Retorno múltiple
# Complete una función que retorne el mínimo y el máximo como una tupla implícita.
def resumen(valores):
    menor = min(valores)
    mayor = max(valores)
    return menor, mayor

minimo, maximo = resumen([8, 3, 10, 5])
print(minimo, maximo)
```

```python
#27. Filtrar diccionarios
# Construya una lista con los nombres de los estudiantes aprobados.
estudiantes = [
    {"nombre": "Ana", "nota": 4.2},
    {"nombre": "Luis", "nota": 2.8},
    {"nombre": "Marta", "nota": 3.5}
]
aprobados = []
for e in estudiantes:
    if e["nota"] >= 3.0:
        aprobados.append(e["nombre"])
print(aprobados)
```

```python
#28. Agrupar por estado
# Agrupe los nombres de los clientes según su nivel de riesgo.
clientes = [
    {"nombre": "Ana", "riesgo": "bajo"},
    {"nombre": "Luis", "riesgo": "alto"},
    {"nombre": "Marta", "riesgo": "bajo"}
]
grupo = {}
for c in clientes:
    riesgo = c["riesgo"]
    if riesgo not in grupo:
        grupo[riesgo] = []     #[]inicia con una lista vacía si el riesgo no existe
    grupo[riesgo].append(c["nombre"])
print(grupo)
```

```python
#29. Inventario académico
# Complete el reporte de los productos con stock bajo.
inventario = [
    {"producto": "Marcador", "stock": 4},
    {"producto": "Cuaderno", "stock": 20},
    {"producto": "Borrador", "stock": 2}
]
bajos = []
for item in inventario:
    if item["stock"] < 20:
        bajos.append(item["producto"])
print(bajos)
```

```python
#30. Mini caso integrados
# Complete el algoritmo que calcula el total de los créditos mayores o iguales a 100000
# y registra los clientes correspondientes.

transacciones = [
    {"cliente": "Ana", "tipo": "Credito", "monto": 120000},
    {"cliente": "Luis", "tipo": "Debito", "monto": 85000},
    {"cliente": "Marta", "tipo": "Credito", "monto": 210000}
]

total_creditos = 0
clientes = []

for t in transacciones:
    if t["tipo"] == "Credito" and t["monto"] >= 100000:
        total_creditos += t["monto"]
        clientes.append(t["cliente"])

print(total_creditos)
print(clientes)

# Indique qué estructuras aparecen en este caso y para qué se usa cada una.
```

***#30 -Estructuras utilizadas y propósito:***
*   Lista (transacciones): almacena una colección ordenada de transacciones.
*   Diccionarios: cada transacción se representa mediante pares clave-valor (cliente, tipo, monto).
*   Lista (clientes): almacena los nombres de los clientes cuyos créditos cumplen la condición.
*   Estructura repetitiva for: recorre todas las transacciones.
*   Estructura condicional if: filtra únicamente las transacciones de tipo crédito con monto mayor o igual a 100.000.
