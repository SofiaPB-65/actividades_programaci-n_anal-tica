**D. Crear información nueva**

Columnas derivadas sin ciclos

```python
##-- 10. Valor de cada unidad vendida
# Cree la columna valor_unitario dividiendo el total entre la cantidad. Resuélvalo sin escribir un ciclo for.

data['valor_unitario'] = data['total'] / data['quantity']
print(data['valor_unitario'].head(3).round(2))
```

```python
##-- 11. Clasificar cada venta
# Marque como volumen las ventas de seis unidades o más y como menor todas las demás.

data['tipo_compra'] = np.where(data['quantity'] >= 6, 'volumen', 'menor')
print(data['tipo_compra'].value_counts())
```

**E. Agrupar para responder preguntas**

El patrón grupo → variable → operación

```python
##-- 12. ¿Cuánto ingreso genera cada sucursal?
# Complete la agrupación que responde la pregunta.

resumen = (
    data
    .groupby('branch')['total']
    .sum()
)
print(resumen.round(2))
```

```python
##-- 13. ¿Cuántas facturas registra cada método de pago?
# Complete la columna sobre la que se cuenta y la operación correspondiente.

resumen = (
    data
    .groupby('payment')['invoice_id']
    .count()
)
print(resumen)

# ¿Qué representa cada fila del resultado agrupado?
# RTA -->
```

```python
##-- 14. Varias métricas por método de pago
# ¿Cuántas facturas, cuántas unidades y cuánto ingreso deja cada método de pago?
# Una sola agrupación debe entregar las tres métricas con nombres explícitos.

resumen = (
    data
    .groupby('payment')
    .agg(
        facturas=('invoice_id', 'size'),
        unidades=('quantity', 'sum'),
        ingreso=('total', 'sum'),
    )
    .reset_index()
)

print(resumen.round(2))
```

**F. Integrar información**

Dos tablas que comparten una clave

```python
##-- 15. Agregar la zona de cada sucursal
# La tabla de ventas no indica en qué zona está cada sucursal.
# Esa información vive en un catálogo aparte. Intégrela conservando todas las ventas.

sucursales = pd.DataFrame({
    'branch': ['A', 'B', 'C'],
    'zona': ['Centro', 'Norte', 'Sur']
})

resultado = data.merge(
    sucursales,
    on='branch',
    how='left'
)

print(resultado.shape)
resultado[['branch', 'city', 'zona', 'total']].head()
```

Descripción punto #16.
Considere únicamente las ventas de clientes de tipo Member. Determine qué sucursal
genera el mayor ingreso total y cuál es su ticket promedio, entendido como el ingreso
dividido entre el número de facturas. Escriba el código completo: aquí no hay espacios
para completar.

```python
##-- 16. La sucursal líder entre los clientes Member

# 1. Filtrar las ventas de clientes Member
# 2. Agrupar por sucursal: numero de facturas e ingreso total
# 3. Calcular la columna ticket_promedio
# 4. Ordenar de mayor a menor ingreso
# 5. Escribir la conclusión

## -------------------------------------

```

```python
## ----
# SOLUCIÓN --> RETO 16
## ----

# 1. Filtrar las ventas de clientes Member
member = data[data['customer_type'] == 'Member'].copy()

# 2. Agrupar por sucursal: número de facturas e ingreso total
resultado = (
    member
    .groupby('branch', as_index=False)
    .agg(
        facturas=('invoice_id', 'count'),
        ingreso_total=('total', 'sum')
    )
)

# 3. Calcular la columna ticket_promedio
resultado['ticket_promedio'] = (
    resultado['ingreso_total'] / resultado['facturas']
)

# 4. Ordenar de mayor a menor ingreso
resultado = resultado.sort_values(
    by='ingreso_total',
    ascending=False
)

# 5. Mostrar resultado
print(resultado)

## --------------------
# --> CONCLUSIÓN
# La sucursal líder presenta el mayor ingreso total entre los clientes Member.
# Su ticket_promedio muestra el valor promedio facturado por transacción.
# Un ticket más alto indica ventas de mayor valor por factura frente a las demás sucursales.
```
